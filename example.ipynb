```python
from solver import QP_solver
import numpy as np
```

```python
# Generate a random non-trivial quadratic program.
n = 2
m1 = 1
m2 = 2

np.random.seed(28)
P = np.random.randn(n, n)
P = P.T @ P
g = np.random.randn(n)
A1 = np.random.randn(m1, n)
b1 = np.random.randn(m1)
A2 = np.random.randn(m2, n)
b2 = A2 @ np.random.randn(n)
```

```python
QP_solver(
    H = P,
    g = g,
    A1 = A1,
    b1 = b1,
    A2 = A2,
    b2 = b2,
    verbose=0,
    plot=True,
    method="ADMM"
)
```

```python
QP_solver(
    H = np.array([[1, -1],
                  [-1, 2]]),
    g = np.array([-2, -6]),
    A1 = None,
    b1 = None,
    A2 = np.array([[2, 2], [-1, 2]]),
    b2 = np.array([-4, -2]),
    verbose=1,
    plot=True,
    method="ADMM"
)
```

```python
QP_solver(
    H = np.array([[1.,-1],
                  [-1, 2]]),
    g = np.array([-2, -6]),
    A1 = None,
    b1 = None,
    A2 = np.array([[1., 1],
                   [-1, 2],
                   [2,  1]]),
    b2 = np.array([-2, -2, -3]),
    verbose=0,
    plot=True,
    method="IRWA"
)
```
